## 2. RDKit Mol Object & Type Verification

In RDKit, chemical structures are represented using SMILES (Simplified Molecular Input Line Entry System) strings. RDKit converts these strings into a **Mol object**, which holds all structural and atomic information.

```python
# 1. Import RDKit Chem library
from rdkit import Chem

# 2. Define example SMILES strings
aspirin_smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"
ethanol_smiles = "CCO"

# 3. Convert SMILES to RDKit Mol Objects
mol_aspirin = Chem.MolFromSmiles(aspirin_smiles)
mol_ethanol = Chem.MolFromSmiles(ethanol_smiles)

# 4. Check and verify Object Types
print("Aspirin Object Type:", type(mol_aspirin))
print("Ethanol Object Type:", type(mol_ethanol))

# 5. Verify validity (Returns 'None' for invalid SMILES)
invalid_mol = Chem.MolFromSmiles("InvalidSMILES123")
print("Invalid SMILES result:", invalid_mol)
```

```python
#Store additional SMILES structure examples in a dictionary (Key: Description, Value: SMILES)
smiles_examples = {
    "Ethanol (Linear Chain)": "CCO",
    "Acetone (Branch & Double Bond)": "CC(=O)C",
    "Cyclopropane (Ring Structure)": "C1CC1",
    "Benzene (Aromatic Ring)": "c1ccccc1",
    "Sodium Acetate (Charged Compound)": "CC(=O)[O-].[Na+]"
}

# . Loop through the dictionary using items()
# 'description' receives the name, 'smiles' receives the SMILES string
for description, smiles in smiles_examples.items():
    # Convert current SMILES to a Mol object
    mol = Chem.MolFromSmiles(smiles)

    # Check if 'mol' is successfully created (not None)
    is_valid = mol is not None

    # Print formatted status for each molecule
    print(f"[{description}] -> Mol Object Created: {is_valid}")
```
