Atom Counting

Get.NumAtoms(),AddHs(),RemoveHs()

```python
smiles = "CCO"  # Ethanol
mol = Chem.MolFromSmiles(smiles)

print("--- Atom Counting Summary ---")
# 1. Count Heavy Atoms (Default)
heavy_atoms = mol.GetNumAtoms()
print(f"Heavy Atoms (C, C, O): {heavy_atoms}")

# 2. Add Explicit Hydrogens
mol_with_hs = Chem.AddHs(mol)
total_atoms = mol_with_hs.GetNumAtoms()
print(f"Total Atoms after AddHs() (C2H6O): {total_atoms}")

# 3. Remove Hydrogens back
mol_no_hs = Chem.RemoveHs(mol_with_hs)
print(f"Atoms after RemoveHs(): {mol_no_hs.GetNumAtoms()}")
```

Grid Drawing (Multiple Molecules)

```python
smiles_list = ["CC(=O)O", "c1ccccc1", "CC(=O)C", "C1CC1"]
labels = ["Acetic Acid", "Benzene", "Acetone", "Cyclopropane"]

mols = [Chem.MolFromSmiles(s) for s in smiles_list]

# Generate a grid image
grid_img = Draw.MolsToGridImage(
    mols,
    molsPerRow=2,
    subImgSize=(250, 250),
    legends=labels
)

display(grid_img)
```
